```python
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
```

```python
df.columns = df.columns.str.lower().str.replace(' ' , '_')
```

```python
df.columns.tolist()
```

```python
df.groupby('iphone_model')['quantity'].sum().sort_values(ascending = False).plot(kind ='bar', figsize=(10,5), title='Units_sold_by_model')
plt.tight_layout()
plt.show()
```

```python
df.groupby('sale_date')['price'].sum().plot(kind='line', figsize=(10,5), title='Revenue_Over_Time')
plt.tight_layout()
plt.show()
```

```python
df.groupby('country')['quantity'].sum().plot(kind='pie',autopct='%1.1f%%',figsize=(7,7), title='Sales_by_Country')
plt.show()
```
